```python
import duckdb
import pandas as pd
import os
from datetime import datetime
```

```python
con = duckdb.connect(database= 'dados_duckdb.db', read_only= False)
```

```python
arquivos = 'z0019_2.csv'
data_ingestao = datetime.now()
df = pd.read_csv (f'../landing/{arquivos}', sep=";")
df['nome_arquvivo'] = arquivos
df['data_ingestao'] = data_ingestao
df.head()
```

```python
print(datetime.now())
```

```python
con.execute ("""
    CREATE  TABLE IF NOT EXISTS bronze_produtos (
             NATBT VARCHAR,
             MAKTX VARCHAR,
             WERKS VARCHAR,
             MAINS VARCHAR,
             LABST VARCHAR,
             nome_arquvivo VARCHAR,
             data_ingestao TIMESTAMP,
        )
""")
```

```python
con.execute("INSERT INTO bronze_produtos SELECT * FROM df")
```
