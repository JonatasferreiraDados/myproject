```python
import duckdb
```

```python
con = duckdb.connect(database = 'dados_duckdb.db',read_only=False)
```

```python
df = con.execute("""
                 SELECT * 
                 FROM(
                    SELECT *, ROW_NUMBER() OVER (PARTITION BY NATBT ORDER BY data_ingestao DESC) AS row
                    FROM bronze_z0019
                    WHERE data_ingestao >='2025-01-20')
                 WHERE row = 1
                 """).fetchdf()
df.head(10)
```

```python
df_final = df.drop(columns=['nome_arquvivo','data_ingestao','row'])
df_final = df_final.rename(columns={"NATBT":"id"})
df_final = df_final.rename(columns={"MAKTX":"nm_produto"})
df_final = df_final.rename(columns={"WERKS":"id_categoria"})
df_final = df_final.rename(columns={"MAINS":"id_fornecedor"})
df_final = df_final.rename(columns={"LABST":"vl_preco"})
df_final.head(10)
```

```python
df_final.dtypes
```

```python
df2 = df_final
df2 = df2.astype(
    {
        'id': 'int32',
        'nm_produto': 'string',
        'id_categoria':'string',
        'id_fornecedor': 'int32',
        'vl_preco': 'float32'
    }

)


df2.head(10)
```

```python
df2 = df_final
df2 = df2.astype(
    {
        'id': int,
        'nm_produto': str,
        'id_categoria':str,
        'id_fornecedor': int,
        'vl_preco': float
    }

)


df2.head(10)
```

```python
con.execute("""
CREATE TABLE IF NOT EXISTS produtos (
        id BIGINT,
        nm_produto TEXT,
        id_categoria TEXT,
        id_fornecedor BIGINT,
        vl_preco FLOAT
        )
""")
```

```python
con.execute("INSERT INTO produtos SELECT * FROM df2")
```

```python
df_resultado = con.execute("SELECT * FROM produtos").fetchdf()
df_resultado.head(10)
```
